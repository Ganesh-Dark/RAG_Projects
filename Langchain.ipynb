```python
#Using template message

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage
from google.colab import userdata
userdata.get('gemini_api_keyy')

api_key = userdata.get("gemini_api_keyy")

llm = ChatGoogleGenerativeAI(
    model="gemma-4-26b-a4b-it",
    temperature=0,
    api_key = api_key
)

messages = [
    SystemMessage(
        content="You are a biology teacher."
    ),

    HumanMessage(
        content="Explain about Viroid"
    )
]

response = llm.invoke(messages)

print(response.content)
```

```python
#chaining

from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from google import genai

from google.colab import userdata
userdata.get('gemini_api_keyy')

api_key = userdata.get("gemini_api_keyy")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a biology teacher."),
    ("human", "{question} who invented that {discovery}")
])

llm = ChatGoogleGenerativeAI(
    model="gemma-4-26b-a4b-it",
    temperature=0,
    api_key = api_key
)

chain = prompt | llm

response = chain.invoke(
    {
        "question": "What is DNA?",
        "discovery" : "DNA"
    }
)

print(response.content)
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/The Odyssey.pdf")

documents = loader.load()

print(type(documents))
```

```python
!pip install -U langchain-community pypdf
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

texts = splitter.split_documents(documents)
#print(texts[0].metadata)


chunk_embeds = []

for chunk in texts:
    chunk_embeds.append(chunk.page_content)

print(len(chunk_embeds))
print(chunk_embeds[0])
```

```python
from langchain_community.vectorstores import FAISS
```

```python
#small = texts[0:10]


db = FAISS.from_documents(
    texts,
    embedding_model
)


```

```python
docs = db.similarity_search(
    "list cast of the odyssey film?",
    k=5
)
```

```python
context = ""

for doc in docs:
    context += doc.page_content + "\n\n"


print(context)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a helpful assistant.

Answer ONLY using the context below.

Context:
{context}

Question:
{question}
""")


final_prompt = prompt.invoke({
    "context": context,
    "question": "list the cast of odyssey and who directed the movie?"
})

from langchain_google_genai import ChatGoogleGenerativeAI

from google.colab import userdata
userdata.get('gemini_api_keyy')

api_key = userdata.get("gemini_api_keyy")

llm = ChatGoogleGenerativeAI(
    model="gemma-4-26b-a4b-it",
    temperature=0, api_key = api_key
)

response = llm.invoke(final_prompt)

print(response.content)
```
